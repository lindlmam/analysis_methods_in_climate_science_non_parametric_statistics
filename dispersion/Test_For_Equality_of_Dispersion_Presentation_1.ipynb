## Intermezzo hypothesis tests - not ranked so far!

**Hypothesis Test:** Are the two distributions are the same or not?
* Null Hypothesis: They are the same  $\rightarrow H_0^*: p_X = p_Y$
* Alternative Hypothesis: They are not the same $\rightarrow H_A^* : p_X \neq p_Y$


If I assume that our samples $X_n$ are **identically and independent distributed (iid)** as $p_x$ and the samples $Y_n$ are iid as $p_y$, we can insert this knowleg in our hypothesis. $\rightarrow$ suppose they are **gaussian aka normally distibuted**. 
>This assumption is also valid for populations which are non-gaussan for a large sample size due to the Central Limit Theorem

Rewrite our hypothese using population parameters of the normal distribution ($\mu$, $\sigma$):
* Null Hypothesis: They are the same $\rightarrow H_0 : \mu_X = \mu_Y$ and $\sigma_X = \sigma_Y$

Since we dont know the real mean $\mu$ and variance $\sigma$, one has to estimate those by the sample mean $\hat{\mu}_X$ and sample variance $\hat{\sigma}_X$.

But what is our alternative hypothesis?

Straightforward approach is to **test equality of means** and **equality of variance** seperately!
Test for equality of variance makes no assuptions about the mean (it uses the sample mean). Thus, tests based on sample variances are noncommittal regarding mean.

$\rightarrow$ Thus it make sens to test equality of variance first! $\rightarrow$ If difference in variance detected $\rightarrow$ difference in distribution.
If no difference in variance detected: assume same variance and test for equality of means



---
**But how to test for difference in variance? F-Test**

The natural statistic for testing equality of variance is the **ratio of variances**!

Let $V_1$ and $V_2$ be independent random variables that are $\chi^2$-distributed with $\nu_1$ and $\nu_2$ degrees of freedom. Then the random variable

$F = \frac{V_1 / \nu_1}{V_2 / \nu_2}$

has a distribution termed the **F-distribution** with $\nu_1$ and $\nu_2$ degrees of freedom.

Assume $X$ and $Y$ are independent and normally distributed, then:

$V_1 = \frac{(N_X-1) \cdot \hat{\sigma}^2_X}{\sigma^2_X}$

$\nu_1 = N_X-1$

$\frac{V_1}{\nu_1} = \frac{\hat{\sigma}_X^2}{\sigma_X^2}$

Thus,

$F = \frac{\hat{\sigma}_X^2}{\hat{\sigma}_Y^2}\frac{\sigma_Y^2}{\sigma_X^2}$

Under the null hypothesis: $H_0: \mu_X = \mu_Y$ and $\sigma_X = \sigma_Y$

$\hat{F} = \frac{\hat{\sigma}_X^2}{\hat{\sigma}_Y^2}$

Thus, a **significant difference in variance corresponds to a value of F different from one**! --> This is the F test!


Assign label $X$ to the variable with the larger sample variance to ensure that $\hat{F} > 1$ and then use the decision rules:

* Reject $H_0$   if   $\hat{F} > F_{\alpha/2,N_X-1,N_Y-1}$
* Accept $H_0$   if   $\hat{F} < F_{\alpha/2,N_X-1,N_Y-1}$

---
But our **null hypothesis involves multible assumptions** (normality and independence)! If our data violates some of those assumptions, than that could lead to a rejection of the null hypothese.

One important assumption is that we assume Gaussian distributed data!

**So what should we do if our data is not normaly distributed**?

# Nonparametric Tests or distribution-free Methods - Statistical tests based on ranks
> nonparametric tests can be applied even if the data are in the form of ranks rather than numerical values


Since the **F-test, as a test for differences in variance, is sensitive to departures from normality**, we need a alternative measurement for variance/dispersion

>One reason for this sensitivity is that variance involves the square of a variable, and squares lead to long-tailed distributions that are sensitive to the tail of the original population.

## Median Absolute Deviation (MAD)
**One measure of dispersion** that has attractive properties for a wide class of non-Gaussian distributions is the **median absolute deviation (MAD)**, defined as:

#### $\mathrm{MAD}[X] = \mathrm{median} \biggl[ \underbrace{\bigl| X - \mathrm{median}[X] \bigr| }_{\Delta X}\biggr]$


#### What is the MAD?

The **Median Absolute Deviation (MAD)** is a **statistical non-parametric measure** that quantifies the **dispersion** or variability of a dataset. It is calculated as the median of the absolute differences between each data point and the median of the dataset.

It can be use to check if two datasets have the same level of variability.

> The dispersion describes how much the values in a dataset deviate from the central tendency, such as the mean or median.

#### When should I use the MAD

* **Outlier detection:** MAD is less influenced by extreme values, making it suitable for identifying outliers in a dataset. It can be used as a robust alternative to the standard deviation in outlier detection methods.

* **Robust estimation:** If your dataset contains outliers or is not normally distributed, MAD can provide a more robust estimate of variability compared to the standard deviation.

* **Comparing variability:** When comparing the variability between different datasets, MAD can be useful as it is not affected by differences in the data distribution or the presence of outliers.

>  "robust" refers to the property of a statistical method or measure to be unaffected or less affected by violations of underlying assumptions or the presence of outliers in the data. 

#### When should I not use the MAD

* **Normal distribution assumption:** MAD may not be the best choice if your data is assumed to follow a normal distribution. In such cases, the standard deviation may be more appropriate.

* **Efficiency in certain cases:** While MAD is robust to outliers, it can be less efficient than the standard deviation when dealing with datasets that do not have a large number of outliers.

* **Parametric methods:** MAD is a non-parametric measure, which means it does not rely on specific assumptions about the data distribution. However, if you have prior knowledge about the distribution and can assume it accurately, parametric methods might provide more precise estimates.

## Squared-rank Test

Consider the two data sets:

#### $\Delta X = |X - \mathrm{median}[X]|$
#### $\Delta Y = |Y - \mathrm{median}[Y]|$

If $Y$ has a greater dispersion than $X$, then $\Delta Y$ should be stochatically dominat over $\Delta X$. Thus, one can apply the **Wilcoxon rank-sum test** to $\Delta X$ and $\Delta Y$ to test for equality of disperson. Since the population medians are not known, one has to approximate them by the sample medians $\hat{m}$:

#### $\Delta X^* = |X - \hat{m}_X|$
#### $\Delta Y^* = |Y - \hat{m}_Y|$

Unfortunately, applying the **Wilcoxon rank-sum test** to $\Delta X^*$ and $\Delta Y^*$ resuts in a depedence on the population, which fortunately gets small for large sample sizes.

However, the **sum of the squared ranks**,

### $R_X^{\prime}=\sum_{n=1}^{N_X} \bigl( \textrm{rank}[\Delta X_n^*]\bigl)^2$,

leads to a more powerful test - the so-called **Squared-rank Test for Equality of Dispersion**.

For that, one has to invoke the **Central Limit Theorem** to acquire the mean and variance of the sum of the squared ranks:

### $\mathbb{E} \Bigl[ R_X^{\prime} \Bigl] = N_X \frac{(N+1)(2N+1)}{6}$

### $\mathrm{var} \bigl[ R_X^{\prime} \bigl] \frac{N_X (N-N_X)}{N-1} \Biggl( \frac{1}{N} \sum^N_{i = 1} \biggl( i^2 - \frac{(N+1) (2N+1)}{6} \biggl)^2 \Biggl)$


Finally, the standardized variable $z_{\textrm{cor}}$ has a distribution,

## $ z_{\textrm{cor}} =  \frac{R_X^{\prime} - \mathbb{E} [R_X^{\prime}] + C}{\sqrt{ \textrm{var} [R_X^{\prime}]}}$,

with the continuity correction $C$,

$C=\left\{\begin{array}{ll} -0.5, & R_X > \mathbb{E}[X] \\
         0.5, & R_X < \mathbb{E}[X]\end{array}\right. .$

# The Problem
A scientist decides to investigate if the variability of precipitation in the southeast United States vary between spring and fall?
He formulates the following hypothesis:

* $H_0$ : The variability of precipitation **does not** vary between spring and fall $\rightarrow \hat{\sigma}_X = \hat{\sigma}_Y \rightarrow$ data is **exchangeable** 
* $H_A$ : The variability of precipitation **does** vary between spring and fall $\rightarrow \hat{\sigma}_X \neq \hat{\sigma}_Y \rightarrow$ data is **not exchangeable** 

To investigate this question, a scientist performs a **test for equality of dispersion**. 

Thus, the scientist,
* 1.) collects data for precipitation in the southeast United States for spring and fall.
* 2.) calculates the absolute deviation – that is, the absolute difference between the value and the median value – for each season seperately
* 3.) pooles both data sets together and ranks them
* 4.) seperates them afterwards again, but with their assigned ranks
* 5.) squares the ranked values and sums them up to get the sum of the squared ranks for each data set

If the variability of precipitation **does not** vary between spring and fall, hence, the data is **exchangeable**, then the scientist can invoke the **Central Limit Theorem**.
Thus, the scientist,
* 6.) calculates the mean of the sum squared ranks for one of the data sets
* 7.) calculates the variance of the sum squared ranks for the same data set
* 8.) calcualtes the continuity correnction for this data set
* 9.) finally calculates the standardized variable $z$ and compares it with a suitable threshold $z_{\alpha/2}$

The scientist will decied as followed:
* Reject $H_0$ if $|z| > z_{\alpha/2}$    
* Accept $H_0$ if $|z| < z_{\alpha/2}$  

---

# The implementation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import statsmodels.api as sm
from scipy.stats import norm
```

### 1.) One collects data for precipitation in the southeast United States for spring and fall.

```python
# Create a dictionary with the data
data = {'Year': [2012, 2013, 2014, 2015, 2016],
        'AMJ': [4.26, 5.46, 4.93, 4.27, 4.21],
        'OND': [2.77, 3.51, 3.73, 5.94, 2.69]}

# Create a pandas dataframe from the dictionary
df = pd.DataFrame(data)

# Set the index of the dataframe to be the "Year" column
df.set_index('Year', inplace=True)

# Print the dataframe
print(df)
```

### 2.) One calculates the deviation – that is, the difference between the value and the median value – for each season seperately.
To do so, one has to calculate the corresponding sample medians $\hat{m}_X$ first:
> Its always a good advise to visualize the data

```python
# Calculate the median for each variable
amj_median = df['AMJ'].median()
ond_median = df['OND'].median()
print("Median of 'AMJ' variable:", amj_median)
print("Median of 'OND' variable:", ond_median)

# Create a scatter plot of the data with lines
plt.plot(df.index, df['AMJ'], label='AMJ', color='blue',marker='o', linestyle='-', linewidth=1)
plt.plot(df.index, df['OND'], label='OND', color='orange', marker='o', linestyle='-', linewidth=1)

# Add horizontal lines for medians
plt.axhline(amj_median, color='blue', linestyle='--', label='AMJ median')
plt.axhline(ond_median, color='orange', linestyle='--', label='OND median')

# Add axis labels and a title
plt.xlabel('Year')
plt.ylabel('Value')
plt.title('Plot of AMJ and OND variables with lines')

# Set the x-ticks to only show the year values
plt.xticks(df.index)

# Add a legend
plt.legend()

# Show the plot
plt.show()
```

Now one can subtract the corresponding sample median $\hat{m}_X$ from each data set to acquire the deviation, according to

$ \Delta X^* = |X - \hat{m}_X|$

```python
# Create a new dataframe with deviations from respective medians
dev_df = pd.DataFrame({'AMJ': df['AMJ'] - amj_median, 'OND': df['OND'] - ond_median})

# Print the dataframe
print('Deviations from respective medians:\n')
print(dev_df)
print('\nWith the corresponding medians:\n')
print(dev_df.median())
```

Next, form the absolute value to get the absolute deviation:

```python
# Create a new dataframe with the absolute values of the deviations from respective medians
dev_absolute_df = np.abs(dev_df)
print('Absolute values of the deviations from respective medians:\n')
print(dev_absolute_df)
print('\nWith the corresponding medians:\n')
print(dev_absolute_df.median())
```

Once again, we have calculated the medians of the constructed data set. Since, the data set are the absolute deviations from the mean, one can identify those as the MAD value:
$MAD[X] = median \biggl[ \underbrace{\bigl| X - median[X] \bigr| }_{\Delta X^*}\biggr]$

If we just apply this formula, one can see that it ends up with the same result: 

```python
# Calculate the MAD for each variable
amj_mad = np.median(np.abs(df['AMJ'] - amj_median))
ond_mad = np.median(np.abs(df['OND'] - ond_median))
print("MAD of 'AMJ' variable:", round(amj_mad,2))
print("MAD of 'OND' variable:", round(ond_mad,2))
```

But what does tha MAD value tell us?

Well, it is a robust measure of variability or dispersion of a set of data.

MAD is similar to the standard deviation, but it is less sensitive to outliers and skewed data. This makes it a useful measure of variability in data sets that contain extreme values or are not normally distributed.

But what does us the difference in the MAD values tell us?

Since we have two different MAD values, the variability or dispersion of one data differs from the other. Specifically, the set with the higher MAD value has more variability or dispersion than the set with the lower MAD value.

> Again, visualization often helps to gain a better understanding

```python
print("MAD of 'AMJ' variable:", round(amj_mad,2))
print("MAD of 'OND' variable:", round(ond_mad,2))

# create subplot with 2 horizontally aligned plots
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 5))


# plot the first subplot on ax1
# Create a scatter plot of the deviation data with lines
ax1.plot(df.index, dev_df['AMJ'], label='AMJ', color='blue',marker='o', linestyle='-', linewidth=1)
ax1.plot(df.index, dev_df['OND'], label='OND', color='orange', marker='o', linestyle='-', linewidth=1)

# Add horizontal lines for medians
ax1.axhline(dev_df['AMJ'].median(), color='black', linestyle='--', label='new median')

# Add axis labels and a title
ax1.set_xlabel('Year')
ax1.set_ylabel('Value')
ax1.set_title('Deviations to their medians')

# Set the x-ticks to only show the year values
ax1.set_xticks(df.index)

# Add a legend
ax1.legend()


# plot the second subplot on ax2
# Create a scatter plot of the absolute deviation data with lines
ax2.plot(df.index, dev_absolute_df['AMJ'], label='AMJ', color='blue',marker='o', linestyle='-', linewidth=1)
ax2.plot(df.index, dev_absolute_df['OND'], label='OND', color='orange', marker='o', linestyle='-', linewidth=1)

# Add horizontal lines for medians
ax2.axhline(dev_absolute_df['AMJ'].median(), color='blue', linestyle='--', label='MAD [AMJ]')
ax2.axhline(dev_absolute_df['OND'].median(), color='orange', linestyle='--', label='MAD [OND]')

# Add axis labels and a title
ax2.set_xlabel('Year')
ax2.set_ylabel('Value')
ax2.set_title('Absolute deviations to their medians')

# Set the x-ticks to only show the year values
ax2.set_xticks(df.index)

# Add a legend
ax2.legend()


# adjust spacing between subplots
plt.subplots_adjust(wspace=0.4)

# show the plot
plt.show()
```

### 3.) One pooles both data sets together and ranks them

```python
rank_df = dev_absolute_df.copy()

# combine the two columns into a single column
combined = rank_df.stack().reset_index()
combined.columns = ['Year', 'Quarter', 'Value']

# calculate the combined ranking
combined['Rank'] = combined['Value'].rank(method='average')
print(combined)
```

### 4.) One seperates them afterwards again, but with their assigned ranks
In addition, we also calculate the rank sum, 

### $r_X^{\prime}=\sum_{n=1}^N \textrm{rank}[\Delta X_n]$

,just to get a feeling for our data

```python
# split the combined column back into separate columns
ranked = combined.pivot(index='Year', columns='Quarter', values='Rank')

# rename the columns with '_rank' suffix
ranked.columns = [col + '_rank' for col in ranked.columns]

# sort the columns in ascending order
ranked = ranked.reindex(sorted(ranked.columns), axis=1)

# display the final DataFrame
print(ranked)

# display the rank sum
print('\nThe sum of the ranks is given by: \n')
print(ranked.sum())
```

> Again, it is a good advise to visualize the results

```python
# display the rank sum
print('\nThe sum of the ranks is given by: \n')
print(ranked.sum())

# melt the dataframe to convert the columns into rows
melted_df = ranked.reset_index().melt(id_vars=['Year'], value_vars=['AMJ_rank', 'OND_rank'])

# create dot plot using seaborn
sns.set(style="ticks")
sns.set_context("paper")
sns.catplot(x='value', y='variable', kind='violin', data=melted_df, height=4, aspect=1.5, color=".9", inner=None)
sns.swarmplot(data=melted_df, x='value', y='variable', size=5)

# set the x and y labels
plt.xlabel('Rank')
plt.ylabel('')

# set the title
plt.title('Ranking by Season')

# show the plot
plt.show()
```

### 5.) One squares the ranked values and sums them up to get the sum of the squared ranks for each data set

### $R_X^{\prime}=\sum_{n=1}^{N_X} \bigl( \textrm{rank}[\Delta X_n^*]\bigl)^2$


```python
# square each value in the dataframe
squared = ranked**2 

# display the final DataFrame
print(squared)

# sum the squared values
sum_of_squares = squared.sum()  

# display the rank sum
print('\nThe sum of the ranks is given by: \n')
print(sum_of_squares)
```

> Again, visualization!

```python
# sum the squared values
sum_of_squares = squared.sum()  

# display the rank sum
print('The sum of the ranks is given by: \n')
print(sum_of_squares)

# melt the dataframe to convert the columns into rows
melted2_df = squared.reset_index().melt(id_vars=['Year'], value_vars=['AMJ_rank', 'OND_rank'])

# create dot plot using seaborn
sns.set(style="ticks")
sns.set_context("paper")
sns.catplot(x='value', y='variable', kind='violin', data=melted2_df, height=4, aspect=1.5, color=".9", inner=None)
sns.swarmplot(data=melted2_df, x='value', y='variable', size=5)

# set the x and y labels
plt.xlabel('Rank')
plt.ylabel('')

# set the title
plt.title('Ranking by Season')

# show the plot
plt.show()
```

### 6.) One calculates the mean (expectaion value) of the sum squared ranks for one of the data sets
Invoking the Central Limit Theorem leads to:

### $\mathbb{E} \Bigl[ R_X^{\prime} \Bigl] = N_X \frac{(N+1)(2N+1)}{6}$

```python
# calculate the amount of samples
number_AMJ = len(ranked['AMJ_rank'])
number_OND = len(ranked['OND_rank'])
number_total = number_AMJ + number_OND

# calculate the expectaion value of the sum square ranks invoking the Central Limit Theorem
expecation_value_sum_of_squared_ranks_AMJ = number_AMJ * (number_total + 1) * (2 * number_total + 1) / 6
expecation_value_sum_of_squared_ranks_OND = number_OND * (number_total + 1) * (2 * number_total + 1) / 6

# display the expectaion value of the sum squared ranks
print("Expectation value of the sum squared ranks regarding 'AMJ':", expecation_value_sum_of_squared_ranks_AMJ)
print("Expectation value of the sum squared ranks regarding 'OND':", expecation_value_sum_of_squared_ranks_OND)
```

### 7.) One calculates the variance of the sum squared ranks for the same data set

### $\mathrm{var} \bigl[ R_X^{\prime} \bigl] \frac{N_X (N-N_X)}{N-1} \Biggl( \frac{1}{N} \sum^N_{i = 1} \biggl( i^2 - \frac{(N+1) (2N+1)}{6} \biggl)^2 \Biggl)$

```python
# Create vector as NumPy array
number_total_count = np.arange(1, number_total+1)

# Calculate result using NumPy functions 
variance_sum_of_squared_ranks_AMJ = number_AMJ * (number_total - number_AMJ) / (number_total - 1) * (1 / number_total) * np.sum((number_total_count**2 - (number_total + 1) * (2 * number_total + 1) / 6)**2)
variance_sum_of_squared_ranks_OND = number_OND * (number_total - number_OND) / (number_total - 1) * (1 / number_total) * np.sum((number_total_count**2 - (number_total + 1) * (2 * number_total + 1) / 6)**2)

# display the variance of the sum squared ranks
print("Variance of the sum squared ranks regarding 'AMJ':", variance_sum_of_squared_ranks_AMJ.round(2))
print("Variance of the sum squared ranks regarding 'OND':", variance_sum_of_squared_ranks_OND.round(2))
```

### 8.) One calcualtes the continuity correnction for this data set

### $C=\left\{\begin{array}{ll} -0.5, & R_X > \mathbb{E}[X] \\
         0.5, & R_X < \mathbb{E}[X]\end{array}\right. .$

```python
# calculate the continuity correction for each case:
if sum_of_squares['AMJ_rank'] > expecation_value_sum_of_squared_ranks_AMJ:
    continuity_correction_AMJ = -0.5
else:
    continuity_correction_AMJ = 0.5
    
if sum_of_squares['OND_rank'] > expecation_value_sum_of_squared_ranks_OND:
    continuity_correction_OND = -0.5
else:
    continuity_correction_OND = 0.5
    
# display the variance of the sum squared ranks
print("Continuity correction of the sum squared ranks regarding 'AMJ':", continuity_correction_AMJ)
print("Continuity correction of the sum squared ranks regarding 'OND':", continuity_correction_OND)
```

### 9.) One finally calculates the standardized variable $z$ and compares it with a suitable threshold $z_{\alpha/2}$
> And finds the suitable threshold in the literature

## $ z_{\textrm{cor}} =  \frac{R_X^{\prime} - \mathbb{E} [R_X^{\prime}] + C}{\sqrt{ \textrm{var} [R_X^{\prime}]}}$ 

```python
# calculate the standardized statistic
standardized_variable_AMJ = (sum_of_squares['AMJ_rank'] - expecation_value_sum_of_squared_ranks_AMJ + continuity_correction_AMJ) / np.sqrt(variance_sum_of_squared_ranks_AMJ)
standardized_variable_OND = (sum_of_squares['OND_rank'] - expecation_value_sum_of_squared_ranks_OND + continuity_correction_OND) / np.sqrt(variance_sum_of_squared_ranks_OND)

print('The standardized statistic variable (AMJ) is: z = ',standardized_variable_AMJ.round(5),'(Tie)')
print('The standardized statistic variable (OND) is: z = ',standardized_variable_OND.round(5))

z_AMJ = standardized_variable_AMJ
z_OND = standardized_variable_OND

pval_AMJ = norm.sf(abs(z_AMJ))*2
pval_OND = norm.sf(abs(z_OND))*2
print('\nP-value = ',pval_AMJ.round(3))
print('P-value = ',pval_OND.round(3))

# Get 5% critical value for a standard normal distribution - since twotaild use 2.5%
critical_value = norm.ppf(0.975)

print('\nThe 5% critical value for a normal distribution is: z_alpha/2 = ', critical_value.round(2))
z_alpha = critical_value

z = z_AMJ
pval= pval_AMJ
```

### Finally, its time for a decision! - One has to decide as followed:
### - Reject $H_0$ if $|z| > z_{\alpha/2}$    
### - Accept $H_0$ if $|z| < z_{\alpha/2}$ 

```python
if abs(z_AMJ) < critical_value:
    print('\nSince, abs(z) < z_alpha/2, the equal dispersion hypothesis cannot be rejected!')
else:
    print('\nSince, abs(z) > z_alpha/2, the equal dispersion hypothesis has to be rejected!')                                             
```

### Thus, the scientist decides to accepts the null hypothesis,

* $H_0$ : The variability of precipitation **does not** vary between spring and fall $\rightarrow \sigma_x = \sigma_y \rightarrow$ data is **exchangeable** 



```python
if abs(z) < z_alpha:
    print('\nSince, abs(z) < z_alpha/2, the equal dispersion hypothesis cannot be rejected!')
    z_line_color = 'green'
else:
    print('\nSince, abs(z) > z_alpha/2, the equal dispersion hypothesis has to be rejected!')     
    z_line_color = 'green'

# Generate x values for the normal distribution plot
x = np.linspace(-4, 4, 1000)

# Calculate the y values for the normal distribution plot
y = norm.pdf(x, 0, 1)

# Create the plot
fig, ax = plt.subplots(figsize=(15, 5))

# Plot the normal distribution curve
ax.plot(x, y, 'k-', linewidth=2, label='Standard Normal Distribution')

# Shade the area corresponding to the right rejection region
x_fill_r = np.linspace(z_alpha, 4, 100)
y_fill_r = norm.pdf(x_fill_r, 0, 1)
ax.fill_between(x_fill_r, y_fill_r, color='red', alpha=0.3, label='Rejection Area')

# Shade the area corresponding to the left rejection region
x_fill_l = np.linspace(-4,-z_alpha, 100)
y_fill_l = norm.pdf(x_fill_l, 0, 1)
ax.fill_between(x_fill_l, y_fill_l, color='red', alpha=0.3)

# Shade the area corresponding to the acceptance region
x_fill_critical = np.linspace(-z_alpha, z_alpha, 100)
y_fill_critical = norm.pdf(x_fill_critical, 0, 1)
ax.fill_between(x_fill_critical, y_fill_critical, color='green', alpha=0.3, label='Acceptance Region')

# Plot the vertical lines for the standardized statistic and critical value
ax.axvline(x=abs(z), color=z_line_color, linestyle='--', label='Standardized Statistic (z)')
ax.axvline(x=z_alpha, color='k', linestyle='--', label='5% Critical Value (z_alpha/2)')

# Set labels and title
ax.set_xlabel('Normal Probability')
ax.set_ylabel('Probability Density')
ax.set_title('Standard Normal Distribution')

# Add legends
ax.legend(loc = 2)

# Add annotations for p-value and critical region
ax.annotate(f'z = {abs(z.round(2))}\np-value = {pval.round(2)}', xy=(abs(z) + 0.1, 0.1), color='red')
ax.annotate(f'Acceptance Region\n           95%', xy=(-1.2, 0.05), color='green')
ax.annotate(f'Rejection Region\n           5%', xy=(-4, 0.05), color='red')

# Show the plot
plt.show()
```
