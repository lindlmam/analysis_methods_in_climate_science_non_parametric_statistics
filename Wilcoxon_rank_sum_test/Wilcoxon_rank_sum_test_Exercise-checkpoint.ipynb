# Wilcoxon rank sum test

## comparing snowheight & temperature of different periods

In this example you'll see how the wilcoxon rank sum test can be applied on climate relevant data. 

This example is about comparing the monthly average snowheight or temperature of two periods in the same location.
The two periods differ in the prevailing temperature. 
Thus the question is, if rising temperatures had a significant impact on the snowcover, or if there was a significant change in temperature.
You'll always look on monthly averages.

The hypothesis you are testing within this example are:

* $H_0$ : The snow height **does not** change between the two periodos $\rightarrow \hat{\sigma}_X = \hat{\sigma}_Y \rightarrow$ data is **exchangeable** 
* $H_A$ : The snow height **does** change between the two periods$\rightarrow \hat{\sigma}_X \neq \hat{\sigma}_Y \rightarrow$ data is **not exchangeable** 

You can also check if there was a significant change in the mean temperature of two periods.
* $H_0$ : Temperature **does not** change between the two periodos $\rightarrow \hat{\sigma}_X = \hat{\sigma}_Y \rightarrow$ data is **exchangeable** 
* $H_A$ : TTemperature **does** change between the two periods$\rightarrow \hat{\sigma}_X \neq \hat{\sigma}_Y \rightarrow$ data is **not exchangeable**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import calendar

from scipy.stats import wilcoxon
from scipy.stats import ranksums
```

## set the filename in the next step:

```python
# at first, choose your file
filename = '/Users/andreaswedenig/Desktop/non_parametric_statistics/InitialData/SHBärental.csv'
```

```python
# set overall lenght, please leave it like this, as not all datasets start at this date.
start_date = '1960-01-01'
end_date = '2016-01-01'
# read csv and store it in pd dataframe 
df_1 = pd.read_csv(filename, sep='\s+', header=1, names = ('Date','Time', 'delim', 'Height'), encoding = "ISO-8859-1")
# make sure that numbers are float not object type
df_1 = df_1.rename(columns={'Height': 'data_1'})
df_1['data_1'] = pd.to_numeric(df_1['data_1'].str.replace(',', '.'))
df_1['Date'] = pd.to_datetime(df_1['Date'], format='%d.%m.%Y')
df_1.set_index('Date', inplace=True)
# filter after predefined period
df_filtered = df_1[start_date : end_date]
df_filtered = df_filtered[["data_1"]]
# resample to monthly frequency
df_monthly = df_filtered.resample('M').mean()
```

## Set the month you want to investigate

* 1 = January ... 12 = December
* consider that there is no snow in some months! 
* you'll have to change only the variable "monthofinterest"

```python
#select month you want to investigate

monthofinterest = 1


month_name = calendar.month_name[monthofinterest]
df_month_mean = df_monthly.loc[df_monthly.index.month == monthofinterest] # 1 = January... 12 = December
# calculate the monthly mean
df_month_mean = df_month_mean.groupby(df_month_mean.index.year).mean()
```

## Have a look on the data you selected

```python
plt.figure(figsize=(0.5, 0.5)) #plot data
df_month_mean.plot(kind='line', color='blue')
plt.xlabel('Year')
plt.ylabel('Snow Height [cm] or Temprerature[C°]')
plt.title(f"{month_name} mean values")
plt.show()
```

## Next you have to define the lenght of the period and the starting year of each period. 
* the lenght(years) is defined in **"periodlenght"**, shall be a anumber between **5...20**
* the starting year is defined in **"startperiod1 & startperiod2"**, shall be a year in format xxxx , for example 1960 and 2000
* please **consider** that the datasets **end in year 2015**, so the length of the period + the starting year of the latter period shall not exceed 2015

```python
# define period length and start years
periodlength = 20
startperiod1 = 1960
startperiod2 = 1995



# filter original dataframe based on start and end years for each period
df_period1 = df_month_mean[(df_month_mean.index >= startperiod1) & (df_month_mean.index < startperiod1 + periodlength)]
df_period2 = df_month_mean[(df_month_mean.index >= startperiod2) & (df_month_mean.index < startperiod2 + periodlength)]
```

## in the next plot you can see the 2 periods you want to perform the rank sum test on

```python
# plot data
plt.figure(figsize=(8, 5))
plt.plot(df_period1.index, df_period1['data_1'], color='blue', linestyle='--', label='Period 1')
plt.plot(df_period2.index, df_period2['data_1'], color='red', linestyle='--', label='Period 2')
plt.xlabel('Year')
plt.ylabel('Snow Height [cm]')
plt.title(f"{month_name} Snow Height for Two Periods")
plt.legend()
plt.show()
```

## In the next step we start with the ranking

* therefore we throw both periods again in one pot and order them ascending
* the smallest value gets rank 1


```python
df_period1 = df_period1.reset_index()
df_period1['Date'] = df_period1.index + 1
df_period1 = df_period1.set_index('Date')
df_period2 = df_period2.reset_index()
df_period2['Date'] = df_period2.index + 1
df_period2 = df_period2.set_index('Date')
df_period1 = df_period1.rename(columns={'data_1': 'values'})
df_period2 = df_period2.rename(columns={'data_1': 'values'})
df_period1and2 = pd.concat([df_period1, df_period2], axis=1)
df_period1and2.columns = ['period_1', 'period_2']
# create a new dataframe with all values
df_ranked = pd.DataFrame({'values': df_period1and2['period_1'].tolist() + df_period1and2['period_2'].tolist()})
# rank the values and add a new column with the ranks
df_ranked['ranks'] = df_ranked['values'].rank(method='min')
df_ranked = pd.DataFrame(df_ranked, columns=['values', 'ranks'])
df_sorted = df_ranked.sort_values('ranks', ascending=True) #sort values to show ascending order
```

## The next step is to split the values again by their period

```python
df_compared1 = pd.merge(df_period1,df_ranked , on='values', how='left')
df_compared2 = pd.merge(df_period2,df_ranked , on='values', how='left')
```

## Let's see how the ranks are distributed!

```python
x1 = df_period1['values']
x2 = df_period2['values']
y1 = 0.5  # Set y values as constant
y2 = 1.5 # Set y values as constant
# Create scatter plot on x-axis only
plt.scatter(x1, [y1]*len(x1), s=40, color='blue', label='Data P1')
plt.scatter(x2, [y2]*len(x2), s=40, color='red', label='Data P2')
plt.title('Comparison of ranks Period & Period 2')
plt.xlabel('ranks')
plt.ylabel('')
plt.ylim(0, 2)
plt.gca().set_yticklabels([])
plt.legend(loc='lower right')
plt.show()
```

## Now we can sum all rank values per period

```python
ranksum1 = df_compared1['ranks'].sum()
ranksum2 = df_compared2['ranks'].sum()
print("Ranksum of P1 is:", ranksum1)
print("Ranksum of P2 is:", ranksum2)
```

With the rank sum and the number of ranks we already have everything we need to perform the wilcoxon rank sum test, therefore we have to: 
* determine the mean
$R_X = \sum_{n=1}^{N_X} \text{rank}[X_n]$
* determine the variance
$E[R_X] = \frac{N_X(N_X+N_Y+1)}{2}$
* and then calculate the z value
$
z_{cor} =\frac{R_X - E[R_X] + C}{\sqrt{\text{var}[R_X]}}$
where 

$C = -0.5$ for $R_X > E[R_X]$

$C = +0.5$ for $R_X < E[R_X]$


## luckily we're working on pretty advanced calculators, there are several pyhton packages that help us in this case. here we will use the wilcoxon tests from the scipy package

```python
df_period2
```

```python
# Extract the values from the two dataframes
x = df_period1['values']
y = df_period2['values']

# Perform the Wilcoxon rank sum test
statistic, p_value = ranksums(x, y)

# Calculate expected value
n1 = len(x)
n2 = len(y)

mean_rank = (n1 * (n1 + n2 + 1)) / 2
var_rank = (n1 * n2 * (n1 + n2 + 1)) / 12

# calculate Z-score
Z = (ranksum1 - mean_rank -0.5) / np.sqrt(var_rank)

# Print the test statistic and p-value
print("length of x:", n1)
print("length of y:", n2)
print("p-value:", p_value)
print(f"Expected rank sum: {mean_rank:.4f}")
print(f"Variance of rank sum: {var_rank:.4f}")
print(f"Z-score: {Z:.4f}")
```

## Let's plot the corrected Z value and see in which region we and up, this will tell us how to interpret our hypthesis!

```python
# Generate data for the normal distribution
mu = 0  # mean
sigma = 1  # standard deviation
x = np.linspace(mu - 4 * sigma, mu + 4 * sigma, 100)
y = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)


fig, ax = plt.subplots()
# Plot the normal distribution curve
ax.plot(x, y, color='black')
rejection_area = ax.fill_between(x, y, where=(np.abs(x) >= 1.96), color='red', alpha=0.5)
acceptance_area = ax.fill_between(x, y, where=((np.abs(x) <= 1.96) & (np.abs(x) >= 0)), color='green', alpha=0.3, step="mid")
ax.set_xlabel('Standard Deviations')
ax.set_ylabel('Probability Density')
ax.set_title('Normal Distribution with p-value')
legend_handles = [rejection_area, acceptance_area, ax.axvline(x=statistic, color='blue', linestyle='dotted'), ax.axvline(x=1.96, color='red', linestyle='dotted')]
legend_handles = [rejection_area, acceptance_area, ax.axvline(x=statistic, color='blue', linestyle='dotted'), ax.axvline(x=-1.96, color='red', linestyle='dotted')]
legend_labels = ['Rejection Region', 'Acceptance Region', f'Z-score: {Z:.4f}', '5% Critical Value (z_alpha/2)']
ax.legend(legend_handles, legend_labels, loc='upper left')
tick_values = np.linspace(min(x), max(x), 9)  # Adjust the number of ticks as needed
ax.set_xticks(tick_values)
ax.set_xticklabels(['{:.2f}'.format(value) for value in tick_values])
plt.show()
```

### now have a look at the curve, if the blue line is located in the rejection region, we can reject our null hypthesis. 
### if the blue line is located in the green are, then ther's no significant difference at the 5% level between our 2 periods.
### vice versa, if the blue line is located in the red are, there is significant difference at the 5% level between our 2 periods
